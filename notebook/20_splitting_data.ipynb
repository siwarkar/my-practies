## Splitting Strategy

```python
from sklearn.datasets import load_boston
```

```python
bs = load_boston()
bs
```

```python
import pandas as pd
# create an input dataframe
df = pd.DataFrame(bs.data,columns=bs.feature_names)
df[:2]
```

```python
bs.target[:2]
```

```python
# Add output 
df['price'] = bs.target
```

```python
df.columns
```

## When we have a single file/dataset the we need to segment the dataset
## into training data and testing data

```python
df.shape
```

```python
# traditional splitting strategies
# 80% for traiing - 20% testing
# 70% for training - 30% for testing
```

```python
# 80% for traiing - 20% testing
#training: input + output
506*.8
```

```python
#testing : input + output
506*.2
```

```python
# separate input and output from df
X = df.drop(columns='price')
y = df.price
```

```python
display(X[:2],y[:2])
```

```python
# when we r not supplying train_size and test_size so theyr set to None
#means for trianing 75% and testing 25%
display(506*.75,506*.25)
```

```python
#import a function to split
from sklearn.model_selection import train_test_split
```

```python
train_test_split(X,y)
```

```python
# Now wil specify 80% for training
train_test_split(X,y,train_size=.8)
```

```python
#unpack all data
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=.2)
```

```python
len(X_train)
```

```python
# default shuffle is True henace it shuffles all data
X_train
```

```python
# if we set shuffle = False
# then it will take data from top to bottom
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=.2,shuffle=False)
```

```python
X_train[:5]
```

```python
# shuffle the data but only ones
# use random_state = 42
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=.2)
```

```python
X_train[:4]
```

```python
# apply algorithm to check score of data
from sklearn.linear_model import LinearRegression
model= LinearRegression()
#training
model.fit(X_train,y_train)
#check score of training data
model.score(X_train,y_train)*100
```

```python
model.predict([[1.2,0,14.3,0,.45,4.5,15,2.1,11.2,344,12,100,9.8]])
```
