```python
#remove linestyle and add marker
plt.plot(day,temp,color='red',linestyle=' ',marker='*',markersize=10)
plt.grid()
```

```python
#Assign the label to a plot + activate legend
plt.plot(day,temp,color='red',linestyle='--',marker='*',markersize=10,label='Temperature')
plt.grid()
plt.legend()#used to make label visible in plot
```

```python
#Assign the label to a plot
# X axis label , Y axis label and Title
plt.plot(day,temp,color='red',linestyle='--',marker='*',markersize=10,label='Temperature')
plt.grid()
plt.legend()#used to make label visible in plot
plt.xlabel('Days',fontsize=15)
plt.ylabel('Temperature per day')
plt.title('Dayswise Temperature')
```

```python
# Save the plot as img
plt.plot(day,temp,color='red',linestyle='--',marker='*',markersize=10,label='Temperature')
plt.grid()
plt.legend()#used to make label visible in plot
plt.xlabel('Days',fontsize=15)
plt.ylabel('Temperature per day')
plt.title('Dayswise Temperature')
plt.savefig('datTemp1.jpg',dpi=300,transparent=True)
```
