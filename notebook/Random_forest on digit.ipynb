```python
import pandas as pd
```

```python
from sklearn.datasets import load_digits
```

```python
digits= load_digits()
```

```python
print(type(digits))
```

```python
print(dir(digits))
```

```python
print(digits.target_names)
```

```python
print(digits.images[0])
type(digits.images[0])
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
plt.gray()
for i in range(2,5):
    plt.matshow(digits.images[i])
```

```python
#lets create dataframe
df = pd.DataFrame(digits.data)
#digits.images[0]
df.head(2)
```

```python
#Now will add target to out digit dataframe
df['target']=digits.target
df.head()
```

```python
X =df.iloc[:,:-1] #independant:64 feature
```

```python
y = df.iloc[:,64] #dependent
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
#from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()
#Default n_estimators=100 means number of Trees
model.fit(X_train,y_train)
```

```python
#check the perfromance
model.score(X_test,y_test)*100
```

```python
#Do the predictive modelling
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred)
cm
```

```python
import seaborn as sns 
plt.figure(figsize=(10,7))
sns.heatmap(cm,annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Plot the confusion matrix in graph
cm = confusion_matrix(y_test,y_pred, labels=model.classes_)

# ploting with labels
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
plt.figure(figsize=(10,7))
disp.plot()

# showing the matrix
plt.show()
```
