```python
plt.scatter(range(10),np.linspace(2,5,10),label='Points')
plt.xlabel('Data')
plt.ylabel('Values')
plt.title('Sample Graph')
plt.legend()
```

```python
# on X axis we need to give new labels
plt.scatter(range(10),np.linspace(2,5,10),label='Points')
plt.xlabel('Data')
plt.ylabel('Values')
plt.title('Sample Graph')
plt.legend()
#set labels on X axis
plt.xticks(range(10),['D12312321','D1231232','D1231233','D4','D5','D6','D7','D8','D9','D10'],
          rotation=45)

# only displays image
plt.show()
```

```python
# Use markers
plt.scatter(range(10),np.linspace(2,5,10),label='Points',marker='*')
plt.xlabel('Data')
plt.ylabel('Values')
plt.title('Sample Graph')
plt.legend()
#set labels on X axis
plt.xticks(range(10),['D12312321','D1231232','D1231233','D4','D5','D6','D7','D8','D9','D10'],
          rotation=45)

# only displays image
plt.show()
```

```python
# change figure size
plt.figure(figsize=(3,4))
plt.scatter(range(10),np.linspace(2,5,10),label='Points')
plt.xlabel('Data')
plt.ylabel('Values')
plt.title('Sample Graph')
plt.legend()
#set labels on X axis
plt.xticks(range(10),['D1','D2','D3','D4','D5','D6','D7','D8','D9','D10'],
          rotation=45)

# only displays image
plt.show()
```

```python
ps = df.Post.value_counts()
ps
```

```python
plt.bar(ps.index,ps.values,color='red')
```

```python
# set different colors
plt.bar(ps.index,ps.values,color=['red','yellow','cyan','magenta','green','black'])
```

```python
# vist: https://html-color.codes/
# set differeent colors using Hexcode

plt.bar(ps.index,ps.values,color=['#8b0000','#b8860b','cyan','magenta','green','black'])
```

```python
new = [8,4,4,2,4,6]
```
