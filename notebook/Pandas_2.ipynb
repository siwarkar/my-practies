## Selection can be performed using 2 other options
- loc
- iloc

## loc: label based selection

```python
#df.loc[]# it will first work on rows
df.loc[3:]
```

```python
df.loc[::-1] #reverse rows
```

```python
#other option in loc
df.loc[:,:] 
# [rows,columns]
```

```python
df.loc[:,'Name':'Post']
# stop is inclusive
```

```python
df.loc[:,'Gender'::2]
```

```python
#access name ,post , salary
df.loc[:,::2]
```

### Lets work on rows

```python
df.loc[::2,'Name':'Age']
```

```python
#harsh,megha,Vrunda,pravin
df.loc[[0,1,5,3],:]
```

```python
# Name,Salary, Post
df.loc[:,['Name','Salary','Post']]
```

## iloc: integer/index based selection

```python
df.iloc[:,:]
```

```python
df.iloc[1:3,:]
# stop is exclusive in iloc
```

```python
#work on columns
df.iloc[:,:2]
```

```python
# column: Post , Name, Age
# row: Tejas,Megha,Harsh
df.iloc[[4,1,0],[2,0,3]]
```

```python
# Access rows present at odd index
df.iloc[1::2]
```

```python
df[1::2]
```
