### np.where()

```python
a
```

### convert values > 10 as 1 and values<=10 as 0

```python
# no.where(condition, return x if True , else y)
np.where(a>10,1,0)
```

```python
# suqare value if its % 5 ==0
np.where(a%5==0,a**2,-a)
```

```python
np.where(a%5==0,a**2,0)
```

```python
condition = np.array(['on','off','on','on','off'])
condition
```

```python
# Convert on as 1 and off as 0
np.where(condition=='on',1,0)
```
