# <font color = blue> <marquee> **Seaborn** </marquee> </font> 

- Seaborn is an open source, BSD-licensed Python library providing high level API for visualizing the data using Python programming language.

- It is built on top of Matplotlib.



**Seaborn Vs Matplotlib**

*It is summarized that if Matplotlib “tries to make easy things easy and hard things possible”, Seaborn tries to make a well-defined set of hard things easy too.”*

*Seaborn helps resolve the two major problems faced by Matplotlib; the problems are −*

1.   Default Matplotlib parameters
2.   Working with data frames

# Plotting using Seaborn

**Histogram using distance plotting**

Histograms represent the data distribution by forming bins along the range of the data and then drawing bars to show the number of observations that fall in each bin.

```python
sns.histplot(x='price',data=df)
```

```python
#check fair hist plot
sns.histplot(x='carat',data=df,bins=[.20,.35,.45,.55])
```

### Combine hist + kde(Kernel density estimation plot)

```python
#Flexibly plot a univariate distribution of observations.
sns.distplot(df.carat)
```

```python
i = sns.load_dataset('iris')
i.head()
```

```python
i.species.unique()
```

```python
sns.distplot(i.sepal_length)
```

```python
#now only i want hist plot so will make kde=False
sns.distplot(i.sepal_length,kde=False)
```

```python
sns.displot(data=i,x='sepal_length')
```

```python
sns.displot(data=i,x='sepal_length',kind='kde')
```

```python
# lets take all the columns in hist plot
sns.histplot(data = i)
```

```python
#set new color palette
import matplotlib.pyplot as plt
sns.histplot(data = i,palette='plasma')
plt.title('Hist plot for 4 columns')
plt.xlabel('Data')
plt.ylabel('Count per column')
```

```python
#use all parameters of seaborn
import matplotlib.pyplot as plt
p = sns.histplot(data = i,palette='plasma')
p.set_title('Sample plot hist')
p.set_xlabel('Data')
p.set_ylabel('Frequency')
```

```python
# Kernel Density Estimation (KDE) is a way to estimate the probability density function 
#of a continuous random variable. 
#It is used for non-parametric analysis.
sns.kdeplot(x = 'petal_length',data=i)
```

# Univariate bivariate multivariate??

**Plotting Bivariate Distribution: Joint Plot**

Bivariate Distribution is used to determine the relation between two variables. 

```python
#It gives scatter plot & its strictly bivariate plot
sns.jointplot(x='sepal_length',y='sepal_width',data=i)
```

```python
#hist plot of sepal width
sns.histplot(i.sepal_width)
```

```python
sns.boxplot(x='sepal_width',data=i)
```

```python
#use kind= hex
sns.jointplot(x='sepal_length',y='sepal_width',data=i,kind='hex')
```

```python
#categoriwise scatter plot
sns.jointplot(x='sepal_length',y='sepal_width',data=i,hue='species')
```

```python
#df
sns.jointplot(x='depth',y='table',data=df,hue='cut')
```
