### **🔹 Stemming: The Rule-Based Trimmer**
> **Adil:** *"Stemming is like using a sword to chop off the endings of words to get the root. It doesn’t always make real words, but it’s fast!"*  

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["running", "flies", "easily", "happily", "better", "studies"]
stems = [stemmer.stem(word) for word in words]
print(stems)
```

> **Harry:** *"Wait! ‘flies’ became ‘fli’ and ‘easily’ became ‘easili’? That looks weird!"*  

> **Adil:** *"Exactly! Stemming follows rules but doesn’t care about meaning. That’s why we also use **Lemmatization**!"*  

---

### **🔹 Lemmatization: The Smart Linguist**
> **Adil:** *"Lemmatization is like a dictionary lookup. It finds the correct root word instead of just cutting off endings."*  

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(word) for word in words]
print(lemmas)
```

['running', 'fly', 'easily', 'happily', 'better', 'study']

> **Harry:** *"Whoa! Now ‘flies’ became ‘fly’, and ‘studies’ became ‘study’! This makes more sense!"*  

> **Adil:** *"Yes! Lemmatization is smarter but takes more processing time than stemming."*  

> **Harry:** *"So, if I need speed, I use stemming. If I need accuracy, I use lemmatization!"*

## **Chapter 3: The Mystery of Text Representation**  

Adil and Harry were sitting in their favorite café, sipping on coffee. Harry had a puzzled look on his face.  

> **Harry:** *"Adil, we’ve cleaned and tokenized the text, but how do machines actually understand words?"*

---

### **🔹 Bag of Words (BoW): Counting Words Like a Boss!**  

> **Adil:** *"Imagine we take a bunch of sentences and count how many times each word appears. That’s Bag of Words!"*  

 **Example Sentences:**  
1️⃣ `"Harry loves learning NLP!"`  
2️⃣ `"Adil is teaching NLP to Harry."`  

> **Harry:** *"So we just count the words? That sounds too easy!"*  

> **Adil:** *"Yes, but it works! Let’s see an example in Python."*  

```python
from sklearn.feature_extraction.text import CountVectorizer

texts = ["Harry loves learning NLP!", "Adil is teaching NLP to Harry."]
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())  # List of words
print(bow_matrix.toarray())  # Word counts
```

> **Harry:** *"Oh! Each sentence became a vector, and numbers show word frequencies!"*  

> **Adil:** *"Exactly! But BoW has a problem—it treats all words as equally important. That’s why we use **TF-IDF!**"*

---

### **🔹 TF-IDF: Finding Important Words!**  

> **Harry:** *"TF-IDF? What’s that?"*  

> **Adil:** *"It stands for **Term Frequency - Inverse Document Frequency**. It finds words that are important in a document but rare in others!"*  

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())  # List of words
print(tfidf_matrix.toarray())  # TF-IDF scores
```

> **Harry:** *"Whoa! Words like ‘Harry’ and ‘NLP’ have different importance in each sentence!"*  

> **Adil:** *"Yes! This helps models focus on **important words** rather than just counting everything!"*  

---

### **🔹 N-grams: Understanding Word Relationships**  

> **Harry:** *"Wait! So far, we’ve treated words separately. But what if words appear together in a meaningful way?"*  

> **Adil:** *"That’s where **N-grams** come in! Instead of looking at single words (Unigrams), we can also consider **pairs of words (Bigrams)** or even **triples (Trigrams)!**"*  

```python
vectorizer = CountVectorizer(ngram_range=(1, 2))  # Unigrams and Bigrams
ngram_matrix = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())  # List of N-grams
print(ngram_matrix.toarray())  # N-gram frequencies
```

> **Harry:** *"Whoa! Now we have **word pairs** like ‘learning NLP’ and ‘teaching NLP’!"*  

> **Adil:** *"Exactly! This helps capture **context** better than just single words."*  

---

### **Summary: What Did We Learn?**  

✅ **Bag of Words (BoW)** → Counts words but ignores meaning.  
✅ **TF-IDF** → Identifies important words by considering how rare they are.  
✅ **N-grams** → Captures **word sequences** like **Bigrams (2 words)** and **Trigrams (3 words).**

### **🔹 Word2Vec: Turning Words into Vectors!**  

> **Adil:** *"Word2Vec is a deep learning-based method that turns words into number vectors while capturing their meanings."*  

 **Example:**  
- "King" and "Queen" should have similar representations.  
- "Paris" and "France" should be related.  
- "Apple" (fruit) and "Apple" (company) should be understood based on context.  

> **Harry:** *"That sounds amazing! But how does it do this?"*

---

### **🔹 CBOW (Continuous Bag of Words) – Predicting a Missing Word**  

> **Adil:** *"CBOW predicts a missing word based on surrounding words."*  

 **Example Sentence:**  
 `"Harry loves ____ learning NLP!"`  

> **Harry:** *"The missing word should be ‘deeply’ or ‘actively’!"*  

 **How It Works?**  
- **Input:** ["Harry", "loves", "learning", "NLP"]  
- **Model predicts:** "deeply"  

```python
from gensim.models import Word2Vec
import gensim.downloader as api

# Load a sample dataset
dataset = api.load("text8")  # A dataset used for training Word2Vec models

# Train Word2Vec with CBOW
model_cbow = Word2Vec(dataset, vector_size=100, window=5, min_count=2, sg=0)

# Find similar words
print(model_cbow.wv.most_similar("learning"))
```

[('education', 0.87), ('studying', 0.85), ('knowledge', 0.83)]


> **Harry:** *"Whoa! ‘Learning’ is related to ‘education’ and ‘studying’! The model understands meanings!"*  

> **Adil:** *"Yes! CBOW is great when we have lots of data and it learns fast!"*  

---

### **🔹 Skip-gram – Predicting Surrounding Words!**  

> **Adil:** *"Now, in Skip-gram, we do the opposite! We take a word and try to predict its surrounding words!"*  

 **Example Sentence:**  
 `"Harry loves deeply learning NLP!"`  
- **Input:** "deeply"  
- **Model predicts:** ["loves", "learning"]  

```python
# Train Word2Vec with Skip-gram
model_skipgram = Word2Vec(dataset, vector_size=100, window=5, min_count=2, sg=1)

# Find similar words
print(model_skipgram.wv.most_similar("NLP"))
```

[('AI', 0.89), ('MachineLearning', 0.88), ('DeepLearning', 0.85)]

> **Harry:** *"Oh! Now ‘NLP’ is connected to ‘AI’ and ‘Machine Learning’!"*  

> **Adil:** *"Exactly! Skip-gram works well even with small datasets and learns rare words better than CBOW!"*  

---

### **CBOW vs Skip-gram: Which One is Better?**  

| Feature  | CBOW | Skip-gram |
|----------|------|-----------|
| Speed | ✅ Faster | ❌ Slower |
| Works Well With | ✅ Large Datasets | ✅ Small Datasets |
| Captures Rare Words | ❌ No | ✅ Yes |
| Good for Context Understanding | ✅ Yes | ✅ Yes |

> **Harry:** *"So, if I have lots of data, I use CBOW, but if I want to learn rare words better, I use Skip-gram!"*  

> **Adil:** *"Exactly! Both are powerful in different ways!"*  


### **Summary: What Did We Learn?**  

✅ **Word2Vec** converts words into meaningful vectors.  
✅ **CBOW** predicts missing words using surrounding words.  
✅ **Skip-gram** predicts surrounding words from a single word.  
✅ **CBOW is fast, while Skip-gram learns rare words better.**

### **🔹Preparing the Data**  

**Example Dataset: Movie Reviews**  
| Review | Sentiment |
|--------|-----------|
| "This movie was amazing! I loved every part of it." | 👍 (Positive) |
| "Worst movie ever. Total waste of time." | 👎 (Negative) |

**Importing Libraries & Data:**  

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample dataset
data = {"Review": ["This movie was amazing!", "Worst movie ever!", "I loved it!", "I hated it!"],
        "Sentiment": ["Positive", "Negative", "Positive", "Negative"]}

df = pd.DataFrame(data)

# Splitting data
X_train, X_test, y_train, y_test = train_test_split(df["Review"], df["Sentiment"], test_size=0.2, random_state=42)

# Converting text into TF-IDF features
vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

> **Harry:** *"Okay! We have reviews and their labels. Now what?"*  

> **Adil:** *"Now, we train a machine learning model!"*  

### **🔹Training a Text Classifier**  

**Using a Naïve Bayes Classifier (Good for Text Data!)**  

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Training the classifier
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

# Making predictions
y_pred = model.predict(X_test_tfidf)

# Checking accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))
```

> **Harry:** *"Wow! The model learned how to classify reviews!"*  

> **Adil:** *"Yes! And we can try more models like **Logistic Regression, SVM, or even Deep Learning!**"*  

### **🔹Testing on New Data**  

**Predicting Sentiments for New Reviews**  

```python
new_reviews = ["The movie was fantastic!", "I hated this film so much."]
new_reviews_tfidf = vectorizer.transform(new_reviews)
predictions = model.predict(new_reviews_tfidf)

for review, sentiment in zip(new_reviews, predictions):
    print(f"Review: {review} → Sentiment: {sentiment}")
```

> **Harry:** *"Whoa! The AI knows which reviews are good or bad!"*  

> **Adil:** *"Yep! And we can fine-tune it with **better data and models!**"*

## **Chapter 6: The Mystery of Part-of-Speech (POS) Tagging**  

Adil and Harry were sitting in the library when Harry looked puzzled while reading a sentence.

---

### **🔹 Part-of-Speech (POS) Tagging – Identifying Word Categories!**  

**POS tagging assigns labels to words:**  
**Noun (NN)** – *dog, car, Harry*  
**Verb (VB)** – *run, eat, code*  
**Adjective (JJ)** – *happy, blue, fast*  

**Example Sentence:**  
`"Harry runs fast."` → `[('Harry', NN), ('runs', VB), ('fast', RB)]`  

**POS Tagging in Python** 

```python
import nltk
from nltk import pos_tag, word_tokenize

# Sample sentence
sentence = "Harry runs fast."

# Tokenization and POS tagging
tokens = word_tokenize(sentence)
pos_tags = pos_tag(tokens)

print(pos_tags)
```

> **Harry:** *"Nice! It figured out that ‘runs’ is a verb and ‘fast’ is an adverb!"*  

> **Adil:** *"Exactly! But how does it know? That’s where **Hidden Markov Models (HMMs)** come in!"*  

---
### **🔹 Hidden Markov Models (HMM) – Predicting POS Tags!**  

> **Adil:** *"HMM is a probabilistic model that predicts the most likely POS tags based on previous words!"*  

 **How It Works?**  
- **States** → POS Tags (NN, VB, JJ)  
- **Observations** → Words in a sentence  
- **Transition Probability** → Likelihood of moving from one POS tag to another  
- **Emission Probability** → Likelihood of a word belonging to a specific POS  

 **Example:**  

| Word  | Previous POS  | Most Likely POS |
|-------|-------------|---------------|
| Harry | ? | **NNP (Proper Noun)** |
| runs  | NNP | **VB (Verb)** |
| fast  | VB | **RB (Adverb)** |

 **HMM POS Tagging in Python**  

```python
import nltk
from nltk.tag import hmm

# Training HMM-based POS tagger
train_data = [[('Harry', 'NNP'), ('runs', 'VBZ'), ('fast', 'RB')],
              [('The', 'DT'), ('dog', 'NN'), ('barks', 'VBZ')]]

trainer = hmm.HiddenMarkovModelTrainer()
hmm_tagger = trainer.train(train_data)

# Testing
print(hmm_tagger.tag(['The', 'dog', 'runs', 'fast']))
```

[('The', 'DT'), ('dog', 'NN'), ('runs', 'VBZ'), ('fast', 'RB')]


> **Harry:** *"Whoa! So, the model learns from labeled sentences and applies it to new ones!"*  

> **Adil:** *"Exactly! But to get the best sequence of POS tags, we use the **Viterbi Algorithm!**"*  


### **🔹 Viterbi Algorithm – Finding the Best POS Sequence!**  

 **What does it do?**  
 Given a sentence, it finds the **best sequence of POS tags**.  
 It **chooses the most probable path** using probabilities from HMM.  

 **Example:**  
For `"The dog runs fast."`, the model considers multiple paths:  
1️⃣ `(DT, NN, VB, RB)`  
2️⃣ `(DT, JJ, VB, RB)`  
3️⃣ `(DT, NN, NN, RB)`  

And **Viterbi selects the most probable one!**  

 **Viterbi Algorithm in Python**  

```python
from nltk.tag import hmm

# Testing with Viterbi algorithm
sentence = ['The', 'dog', 'runs', 'fast']
best_tags = hmm_tagger.tag(sentence)
print(best_tags)
```

[('The', 'DT'), ('dog', 'NN'), ('runs', 'VBZ'), ('fast', 'RB')]

> **Harry:** *"So, Viterbi helps the model find the best possible POS sequence!"*  

> **Adil:** *"Exactly! It ensures the highest probability path is chosen!"*  

### **🔹 Summary: What Did We Learn?**  

✅ **POS Tagging** helps machines understand grammar.  
✅ **HMMs** predict POS tags based on probabilities.  
✅ **Viterbi Algorithm** finds the best POS sequence.
