## Problem Statement: Identify either note is Genuine or not using predictive modelling  and with the help of Decision tree

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('banknotes.csv')
df.head()
```

```python
# lets Create input and output data
X = df.drop(columns='Class')
X[:3]
```

```python
y = df['Class']
y[:3]
```

```python
# splitting the dataset
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2)
```

```python
# Import Descision tree algo
#from sklearn.tree import DecisionTreeRegressor

# current problem is binary classification problem
# hence classifier wil be used
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
model.classes_
```

```python
# prediction
y_pred = model.predict(X_test)
y_pred[:5]
```

```python
y_test[:5]
```

```python
# Lets check trianing data score
model.score(X_train,y_train)*100
```

```python
model.score(X_test,y_test)*100
```

```python
#performance metrics
from sklearn.metrics import confusion_matrix,classification_report
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
model.criterion
```

```python
# LEst apply entropy
model2 = DecisionTreeClassifier(criterion='entropy')
```

```python
model2.fit(X_train,y_train)
```

```python
y_pred2 = model2.predict(X_test)
```

```python
confusion_matrix(y_test,y_pred2)
```

```python
print(classification_report(y_test,y_pred2))
```
