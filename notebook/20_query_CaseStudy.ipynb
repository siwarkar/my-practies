## query: evaluates expression and returns df

```python
#Query using df

df = pd.DataFrame({'A': range(1, 6),
                   'B': range(10, 0, -2),
                   'C': range(10, 15)})
df
```

```python
#Query the columns of a DataFrame with a boolean expression
df.query('A > B')
```

```python
df.query('A==3')
```

```python
# more than one condition
out = df.query('B>4 and A>1')
out[['A','C']]
out.loc[:,['C','A']]
out.iloc[:,[0,2]]
```

```python
#fetch the value from column B where value of column A==3
df.query('A==3').B
```

```python
#fetch columb B where A is 3
b1 = df.query('A==3')
b1['B']
```

```python
# answer of above question using indexing
df.A == 3
```

```python
df[df.A == 3]
```

```python
# fetch column A and C where B value is =10
j1 = df[df.B == 10]
j1[['A','C']]
```

```python
df[df.B == 10][['A','C']]
```

```python
#using query
df.query('B == 10')[['A','C']]
```
