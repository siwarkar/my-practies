## Creation of df

```python
# list
d = [10,20,30,40]
pd.DataFrame(d)
```

```python
pd.Series(d)
```

```python
# list of list
k = [[1,2],[3,4]]
k
```

```python
pd.DataFrame(k)
```

```python
# Lets change index and column name
pd.DataFrame(k,index=['A','B'],columns=['Data_1','Data_2'])
```

```python
# positional arguments
pd.DataFrame(k,['A','B'],['Data_1','Data_2'])
```

```python
# but if we interchange positions
print(pd.DataFrame([1,2,3],['A','B','C']))
print(pd.DataFrame(['A','B','C'],[1,2,3]))
#it will change output
```

```python
# it accepts homo. Hetro values
pd.DataFrame([10,23.4,'A','Amit'])
```

```python
# Using set
s = {(1,2,3),(1,2,5)}
# one tuple is one row
pd.DataFrame(s)
```

```python
# Using Dict
d = {'name':['Ashok','Seema'],'age':[23,24],'place':['pune','sangli']}
pd.DataFrame(d)
# key become a column label
# values will be added with respect to keys
```

```python
# Using Series
s = pd.Series([120,300,500,750])
pd.DataFrame(s)
```

```python
# Using numpy
import numpy as np
n = np.random.random(5)
pd.DataFrame(n)
```

```python
# use 2D array to create a dataframe
n2 = np.random.random((5,4))
pd.DataFrame(n2)
```
