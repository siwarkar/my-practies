# Plotting Using Pandas

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = np.random.random((5,4))
data
```

```python
#create a dataframe using data
df = pd.DataFrame(data,columns=['a','b','c','d'])
df
```

## lets start plotting

```python
df.b
```

```python
# Line plot:default
df.b.plot()
# index on X axis
# values on y axis
```

```python
#bivariate line plot
#columb b vs column d
df.plot(x='b',y='d')
```

```python
#multivariate plot
#gives plot of all columns
df.plot()
```

```python
#only 2 columns required
df[['a','b']].plot()
```

# Scatter plot

```python
#single column
df.c
```

```python
#bivarriate
df.plot(x='c',y='d',kind='scatter')
```

```python
df.plot.scatter(x='c',y='d')
```

```python
df.plot.scatter()
#as its bivariate, 2 inputs required
```

## Bar plot

```python
df.a
```

```python
df.a.plot(kind='bar')
#on x axis: index
# values of Y axis
```

```python
df.plot(kind='bar')
plt.xticks(range(5),['data1','data2','data3','data4','data5']);
plt.xlabel('Data')
plt.ylabel('Values')
plt.title('Sample.Bar plor')
```

```python
#stack the multiple bar plot
df.plot(kind='bar',stacked=True)
```

```python
df[['c','d']].plot(kind='bar')
```

```python
#subplots
df.plot(kind='bar',subplots=True)
plt.tight_layout()
plt.show()
```

```python
#change shape of plot
df.plot(kind='bar',subplots=True,layout=(2,2))
plt.tight_layout()
plt.show()
```

```python
#change figure size
df.plot(kind='bar',subplots=True,layout=(2,2),figsize=(12,7))
plt.tight_layout()
plt.show()
```

## Horizontal bar plot

```python
df.plot(kind='barh')
```
