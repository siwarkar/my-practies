```python
import numpy as np

x1 = np.linspace(2,4*np.pi,400)
y1 = np.cos(x1)


x2 = np.arange(10)
y2 = np.sin(x2**2)
```

```python
#2x1
plt.subplots(2,1)
```

```python
fig,ax = plt.subplots(2,1)
```

```python
fig
```

```python
ax #it contain plot object
```

```python
# now add values to plot

fig,ax = plt.subplots(2,1)
ax[0].scatter(x1,y1)
ax[1].plot(x2,y2,'--or')
```

```python
# add label to individual plots

fig,ax = plt.subplots(2,1)
ax[0].scatter(x1,y1)
ax[0].set_title('Graph_1')


ax[1].plot(x2,y2,'--or')
ax[1].set_title('Graph_2')

plt.tight_layout() # to justify the spaces between the plots ,it helps to put a gap
```

```python
# Unpcaking of 2 plots as ax1,ax2

fig,(ax1,ax2) = plt.subplots(2,1)
ax1.scatter(x1,y1)
ax1.set_title('Graph_1')
ax1.set_xlabel('Data_1')
ax1.set_ylabel('A')

ax2.plot(x2,y2,'--or')
ax2.set_title('Graph_2')

plt.suptitle('Combining the 2 plot')

plt.tight_layout() # to justify the spaces between the plots ,it helps to put a gap
```

```python
fig,ax= plt.subplots(2,2)
ax[1,1].plot(x1,y1,'--m')#-- line style with m: magenta color

ax[1,0].plot(x2,y2,'-.k')#-.  linestyle and black color
# shortcut only works in line plot
```

```python
fig,(ax1,ax2) = plt.subplots(2,1,sharex=True)
ax1.scatter(x1,y1)
ax1.set_title('Graph_1')

ax2.plot(x1,y1,'--or')
plt.suptitle('Combining the 2 plot')

plt.tight_layout()
```
