# KNN- K nearest Neighbors

```python
# input: [sweetness,crunchiness]
X = [[10,9],[1,4],[10,1],[7,10],[3,10],[1,1],[5,10]]
```

```python
y = ['fruit','protein','fruit','vegetable','vegetable','protein','vegetable']
```

```python
import numpy as np
np.unique(y)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(metric='euclidean')
#algorithm='brute',n_neighbors=3,n_jobs=10
```

```python
model.fit(X,y)
```

```python
tomato = [[6,10]]
```

```python
model.predict(tomato)
```

```python
pineapple = [[10,2]]
model.predict(pineapple)
```

```python
model.score(X,y)*100
```

```python
model.n_neighbors
```
