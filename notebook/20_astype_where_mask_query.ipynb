## where

```python
d>3
```

```python
#Replace values where the condition is False.
d.where(d>3)
```

```python
d.where(d[0]>3)
#where the condition is false there u will get NaN
```

```python
d.where(d[0]>3,0) #if we want to fill other values instead of nan then 
#we can put like this
# so where the condition is False u will get 0
```

```python
d[0].mean()
```

```python
d.where(d[0]>3,d[0].mean())
```

```python
round(12.545556,2)
```

```python
ot = round(d[0].mean(),2)
d.where(d[0]>3,ot)
```

```python
d.where(d<3,-d)
# in above case we are trying to make values negative where condition is False
```

```python
rn = np.random.rand(4,4)
f = pd.DataFrame(rn)
f
```

```python
f.where(f>0.5)
```

```python
f.where(f>0.5,-10)
```

```python
f.where(f>0.5,-f)
```

```python
k = [1,0,0,1,1,0,1]
k
```

```python
s = pd.Series(k)
s.map({1:0,0:1})
```

```python
ar = np.array(k)
ar
```

```python
np.where(ar>0,0,1)
```

```python
fn = lambda x: 0 if x > 0 else 1
```

```python
list(map(fn,k))
```
