```python
import numpy as np
import matplotlib.pyplot as plt
```

## Hist plot
- univariate plot 
- used to check rangewise distribustion
- used to check outliers
- it gives u a data distribustion


```python
#lets create age data
age = np.random.randint(18,80,100)
age
```

```python
plt.hist(age)
```

```python
# supply our range
plt.hist(age,range=(25,60))
```

```python
# specify artists object size
plt.hist(age,range=(25,60),bins=5)
```

```python
# specify own distribustion pattern
plt.hist(age,bins=[20,30,50,60])
```

```python
#change color
plt.hist(age,bins=[20,30,50,60],color='r',alpha=.4)
```

# Pie plot
- Univariate plot
- Out of 100% , gives u %wise distribustion

```python
party = ['BJP','Congress','NCP','Shivsena']
```

```python
count = [100,90,45,80]
```

```python
sum(count)
```

```python
plt.pie(count,autopct='%.2f%%',labels=party)
plt.show()
```

```python
import pandas as pd
d = pd.Series([120,160],index=['Male','Female'])
d
```

```python
d.values
```

```python
plt.pie(d.values,autopct='%.2f%%',labels=d.index)
plt.show()
```

```python
plt.pie(count,autopct='%.2f%%',labels=party,explode=[0,0.2,0,0.4])
plt.show()
```
