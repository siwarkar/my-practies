```python
# as we have 2 classes in the output so will select bernouli NB
from sklearn.naive_bayes import BernoulliNB
```

```python
bnb = BernoulliNB()
```

```python
# Split the Dataset
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2)
```

```python
# Training
bnb.fit(X_train,y_train)
```

```python
#check training data accuracy
bnb.score(X_train,y_train)*100
```

```python
# if we change parameter of Bernouli then will the affect score???
bnb_2 = BernoulliNB(binarize=None)
```

```python
bnb_2.fit(X_train,y_train)
```

```python
bnb_2.score(X_train,y_train)*100
```

```python
# why nt to think of Guasian bcz we have all input features 
#as a contineous values
from sklearn.naive_bayes import GaussianNB
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X_train,y_train)
```

```python
gnb.score(X_train,y_train)*100
```

```python
gnb.score(X_test,y_test)*100
```

```python
# Will go with gnb
# DO the Testing of data
y_pred = gnb.predict(X_test)
y_pred
```

```python
bnb.predict(X_test)# u can see here it give all predictions for class 1
```

```python
y_test #actual test output
```

```python
# As its classification problem so we hv different accuracy measure
"""
Confusion Matrix: TP,TN,FP,FN
Precision
Recall
F1 score
accuracy
Support
TPR
TNR
Specificity
Weighted average
"""
from sklearn.metrics import confusion_matrix,classification_report
```

```python
# this is used to calculate error/residual
# hence we need here, actual output and predicted output
cm = confusion_matrix(y_test,y_pred)
cm
```

```python
sns.heatmap(cm, annot=True,xticklabels=['M','B'],
    yticklabels=['M','B'])
plt.xlabel('Predicted Output')
plt.ylabel('Actual output')
```

```python
# check M:0 records from actual
sum(y_test == 0)
```

```python
print(classification_report(y_test,y_pred))
```
